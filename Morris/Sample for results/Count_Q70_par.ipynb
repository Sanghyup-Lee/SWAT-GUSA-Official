```python
import pandas as pd
import os

# 경로
base_path = r"C:\Users\sl177\OneDrive - University of Illinois - Urbana\02_Paper Work\01_Paper(CANOPY)\01_SWAT Uncertainty\Figurework (Results)\Result 2 (Morris)\Sample for results"

quantile_file = os.path.join(base_path, "Quantile70.txt")
parnum_file = os.path.join(base_path, "ParNum.txt")

# ParNum 읽기
par_df = pd.read_csv(parnum_file, sep="\t")
par_list = par_df["ParNum"].tolist()

# Quantile70 읽기
q_df = pd.read_csv(quantile_file, sep="\t", header=None)

# Quantile70 구조
# 0: year
# 1: avg
# 2: Quantile70_Group
# 3: group name
# 4: count
# 5: parameter list (comma separated)

q_df.columns = ["Year", "Type", "GroupTag", "GroupName", "N", "ParList"]

# 파라미터 리스트를 list로 변환
q_df["ParList"] = q_df["ParList"].astype(str).apply(lambda x: [p.strip() for p in x.split(",")])

# 그룹 이름 목록
groups = q_df["GroupName"].unique()

# 결과 저장할 데이터프레임 만들기
result = pd.DataFrame({"ParNum": par_list})

# 그룹별 등장 횟수 카운트
for g in groups:
    subset = q_df[q_df["GroupName"] == g]
    
    counts = []
    for par in par_list:
        # 해당 그룹 subset에서 par이 포함된 row 개수 세기
        cnt = subset["ParList"].apply(lambda plist: par in plist).sum()
        counts.append(cnt)
    
    result[g] = counts

# ParName 붙이기
result = result.merge(par_df, on="ParNum", how="left")

# 컬럼 순서 정리
cols = ["ParNum", "ParName"] + [g for g in groups]
result = result[cols]

# 저장
out_file = os.path.join(base_path, "Quantile70_Counts_by_Group.csv")
result.to_csv(out_file, index=False)

print("완료! 저장 위치:", out_file)
print(result.head(10))
```
